```python
text_doc=loader.load()
#load
# a load >>>>>> asynchronuos doc ko read krta hai 
# lazy load >>>>>> list k data ko iteration krta hai
# alazy load >>>>>>>> asynchronuos list iterate krta hai 
# loader
```

```python
print(text_doc)
```

```python
for doc in text_doc:
    print(doc.page_content)
```

| Function       | Meaning                   | How it works                                     | Return Type      |
| -------------- | ------------------------- | ------------------------------------------------ | ---------------- |
| `load()`       | Load all documents        | Loads the complete data at once                  | `list[Document]` |
| `aload()`      | Async load all documents  | Asynchronously loads all documents at once       | `list[Document]` |
| `lazy_load()`  | Lazy load documents       | Loads documents one by one when needed           | Iterator         |
| `alazy_load()` | Async lazy load documents | Asynchronously loads documents one by one        | Async Iterator   |
| `loader`       | Loader object             | Represents the document loader used to load data | Loader instance  |

```python
import asyncio
from langchain_community.document_loaders import TextLoader
async def main():
    loader=TextLoader("hello.txt")
    doc= await loader.aload()
    print(len(doc))
await main()
```

```python
import asyncio
import nest_asyncio

from langchain_community.document_loaders import TextLoader

nest_asyncio.apply()

async def main():
    loader = TextLoader("hello.txt")

    doc = await loader.aload()

    print(len(doc))

asyncio.run(main())
```

```python
import asyncio
import nest_asyncio

from langchain_community.document_loaders import TextLoader

nest_asyncio.apply()

async def main():
    loader = TextLoader("hello.txt")

    doc = loader.alazy_load()

    async for docs in doc:
        print(docs)

asyncio.run(main())
```

```python
import asyncio
import nest_asyncio

from langchain_community.document_loaders import TextLoader,PyPDFLoader

nest_asyncio.apply()

async def main():
    loader = PyPDFLoader("web development report.pdf")

    doc = loader.alazy_load()

    async for docs in doc:
        print(docs)

asyncio.run(main())


```
